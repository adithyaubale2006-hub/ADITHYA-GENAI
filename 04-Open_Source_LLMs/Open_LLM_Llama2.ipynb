### **LAMMA 2**

The Llama 2 is a collection of pretrained and fine-tuned generative text models, ranging from 7 billion to 70 billion parameters, designed for dialogue use cases.

It outperforms open-source chat models on most benchmarks and is on par with popular closed-source models in human evaluations for helpfulness and safety.

meta-llama/Llama-2-13b-hf


llama.cpp's objective is to run the LLaMA model with 4-bit integer quantization on MacBook. It is a plain C/C++ implementation optimized for Apple silicon and x86 architectures, supporting various integer quantization and BLAS libraries. Originally a web chat example, it now serves as a development playground for ggml library features.

### **2: Import all Libraries**

```python
model_name_or_path = "TheBloke/Llama-2-13B-chat-GGUF"
model_base_name = "llama-2-13b-chat.Q5_K_M.gguf"
```

```python
from huggingface_hub import hf_hub_download #Helps to download any particular model from huggingface
from llama_cpp import Llama
```

### **3: Download the model**

```python
model_path = hf_hub_download(repo_id=model_name_or_path, filename=model_base_name)
```

```python
model_path
```

### **4: Loading the Model**

```python
#GPU
lcpp_llm = None
lcpp_llm = Llama(
    model_path,
    n_threads=2, # CPU cores
    n_batch=512, # Should be between 1 and n_ctx, consider the amount of VRAM in your GPU.
    n_gpu_layers=32 # Change this value based on your model and your GPU VRAM pool
)
```

```python
#See the number of layer in Gpu
print("GPU Layers:", lcpp_llm.model_params.n_gpu_layers)
```

### **5: Create A prompt Template**

```python
prompt = "Write a MNIST image classification code in keras"
prompt_template=f'''SYSTEM: You are a helpful, respectful and honest assistant. Always answer as helpfully.

USER: {prompt}

ASSISTANT:
'''
```

### **6: Generating Response**

```python
#here we create an LLM like We did in Openai and Gemini
response = lcpp_llm(
    prompt=prompt_template,
    max_tokens=256,
    temperature=0.5,
    top_p=0.95,
    repeat_penalty=1.2,
    top_k=50,
    echo=True
)
```

```python
print(response['choices'][0]['text'])
```
