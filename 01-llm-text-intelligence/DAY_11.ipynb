````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# 1. Manually load the model and tokenizer to bypass the pipeline registry error
model_name = "sshleifer/distilbart-cnn-12-6"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

text = """
The Eiffel Tower is 324 metres (1,063 ft) tall, about the same height as an 81-storey building, and the tallest structure in Paris. Its base is square, measuring 125 metres on each side. During its construction, the Eiffel Tower surpassed the Washington Monument to become the tallest man-made structure in the world, a title it held for 41 years until the Chrysler Building in New York City was finished in 1930. It was constructed from 1887 to 1889 as the centerpiece of the 1889 World's Fair and was initially criticized by some of France's leading artists and intellectuals for its design, but it has become a global cultural icon of France and one of the most recognizable structures in the world.
"""

# 2. Tokenize and generate the summary
inputs = tokenizer(text, return_tensors="pt", truncation=True)
summary_ids = model.generate(inputs["input_ids"], max_length=50, min_length=20, do_sample=False)
summary_text = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

# 3. Print the result
print(summary_text)
````

#### **SEQ2SEQ MODEL**

````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# 1. Manually load the model and tokenizer to bypass the pipeline registry error
model_name = "Helsinki-NLP/opus-mt-en-fr"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# 2. Define your input sequence
english_text = "Generative AI is changing the world very quickly. I am learning how to build these models."

# 3. Perform the translation
print("--- Seq2Seq Translation (Manual Load) ---\n")
inputs = tokenizer(english_text, return_tensors="pt")
outputs = model.generate(**inputs)
french_translation = tokenizer.decode(outputs[0], skip_special_tokens=True)

print(f"Original (Input Sequence): {english_text}")
print(f"Translated (Output Sequence): {french_translation}")
````

```
                      +-----------------------------+
                      |      Input Sequence (X)     |
                      |   (e.g., "Hello world")     |
                      +--------------+--------------+
                                     |
                                     v
                      +-----------------------------+
                      |           ENCODER           |
                      |   -----------------------   |
                      | Reads entire input sequence  |
                      | and compresses meaning.      |
                      +--------------+--------------+
                                     |
                                     v
                      +-----------------------------+
                      | Context Vector (Thought State)|
                      |   -----------------------   |
                      | A compressed mathematical   |
                      | representation of meaning.  |
                      +--------------+--------------+
                                     |
                                     v
                      +-----------------------------+
                      |           DECODER           |
                      |   -----------------------   |
                      | Uses context vector to      |
                      | generate output step-by-step|
                      +--------------+--------------+
                                     |
                                     |  (Step-by-step generation:
                                     |   y1, then y1+y2, then y1+y2+y3, etc.)
                                     v
                      +-----------------------------+
                      |     Output Sequence (Y)     |
                      |   (e.g., "Bonjour monde")   |
                      +-----------------------------+
```
