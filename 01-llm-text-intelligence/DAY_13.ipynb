#### **LLM FUNDAMENTALS**

**Large Language Models (LLMs)** are advanced AI systems trained on massive amounts of text data to understand patterns and generate human-like language. At their core, they are highly complex "next-word predictors."

**Key Concepts:**

*   **Tokens & Context Window:**
    *   **Tokens:** LLMs don't read raw words; they read "tokens." A token can be a whole word (like "apple"), a part of a word (like "un-"), or even a single character. As a rule of thumb, 1 token is roughly equivalent to 0.75 words.
    *   **Context Window:** This is the model's "short-term memory limit" for a single prompt or conversation. If a model has a 4,000-token context window, it can only "remember" roughly 3,000 words of your prompt and its own response combined. Anything older than that gets forgotten.

*   **Generation Parameters:** These are settings you adjust at inference time (when generating text) to control *how* the model behaves and chooses its next words.

*   **Temperature:** Controls the "creativity" or randomness of the output.
    *   `Low (e.g., 0.1 - 0.3):` The model almost always picks the most likely next word. The output is very focused, predictable, and factual (great for coding or strict Q&A).
    *   `High (e.g., 0.8 - 1.5):` The model takes more risks and picks less common words. The output is more creative, diverse, and sometimes chaotic (great for brainstorming or storytelling).

*   **Top-P (Nucleus Sampling):** Another parameter to control randomness. Instead of tweaking the raw probabilities like Temperature, Top-P restricts the model's choices to only a specific "pool" of likely words. For example, `top_p=0.9` means the model will only consider the top words that make up 90% of the probable choices, completely ignoring the extremely weird/unlikely 10% at the bottom.

````python
from transformers import pipeline, set_seed, logging

# Hide warnings
logging.set_verbosity_error()

# Initialize the text generator
generator = pipeline('text-generation', model='gpt2')

prompt = "The secret to a successful career in tech is"

print("--- 🥶 TEST 1: Low Temperature (0.2) ---")
set_seed(42) # Keeps the starting point consistent for comparison
low_temp = generator(prompt, max_new_tokens=30, temperature=0.2, do_sample=True, pad_token_id=50256)
print(low_temp[0]['generated_text'] + "\n")


print("--- 🥵 TEST 2: High Temperature (1.5) ---")
set_seed(42)
high_temp = generator(prompt, max_new_tokens=30, temperature=1.5, do_sample=True, pad_token_id=50256)
print(high_temp[0]['generated_text'] + "\n")


print("--- 🎯 TEST 3: Top-P (0.8) ---")
set_seed(42)
# Here we use standard temperature (1.0) but restrict the word pool to the top 80%
top_p_test = generator(prompt, max_new_tokens=30, top_p=0.8, do_sample=True, pad_token_id=50256)
print(top_p_test[0]['generated_text'])
````
