#### **PROMPT ENGINEERING**

**Prompt Engineering** is the art and science of communicating effectively with Large Language Models (LLMs) to get the exact output you want. It's like writing clear instructions for a very smart but literal-minded assistant.

**Key Concepts:**

*   **System vs. User Prompts:**
    *   **System Prompt:** The "behind-the-scenes" instructions that set the model's core behavior, persona, and strict boundaries (e.g., "You are a helpful coding assistant. Only answer in Python.").
    *   **User Prompt:** The actual question or task submitted by the human user (e.g., "Write a function to reverse a string.").

*   **Zero-Shot Prompting:**
    Asking the model to perform a task without giving it any examples. You rely entirely on its pre-existing knowledge to figure out what you want.
    *   *Example:* "Classify this text as positive or negative: 'I hated the food.'"

*   **Few-Shot Prompting:**
    Giving the model a few examples of the desired input and output before asking it to complete the final task. This drastically improves accuracy and forces the model to follow a specific pattern or format that it might struggle to understand from instructions alone.

````python
from transformers import pipeline, set_seed, logging

# Hide warnings
logging.set_verbosity_error()

# Initialize standard GPT-2
generator = pipeline('text-generation', model='gpt2')

print("--- 0️⃣ ZERO-SHOT PROMPT ---")
# Asking the model to guess the pattern with no examples
zero_shot_prompt = "Opposites: Hot -> Cold, Up ->"
set_seed(42)
result_zero = generator(zero_shot_prompt, max_new_tokens=5, pad_token_id=50256, temperature=0.5)
print(result_zero[0]['generated_text'] + "\n")


print("--- 🔢 FEW-SHOT PROMPT ---")
# Giving the model a clear pattern to follow first
few_shot_prompt = """Opposites:
Big -> Small
Fast -> Slow
Hot -> Cold
Up ->"""
set_seed(42)
result_few = generator(few_shot_prompt, max_new_tokens=5, pad_token_id=50256, temperature=0.5)
print(result_few[0]['generated_text'])
````

````python
import os
from openai import OpenAI

# 1. Initialize the client
# Replace with your actual OpenAI API key, or use Colab Secrets
os.environ["OPENAI_API_KEY"] = "your-api-key-here"
client = OpenAI()

print("--- 🎭 SYSTEM VS USER PROMPT ---")
# The System message sets the rules, the User message asks the question
role_response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[
        {"role": "system", "content": "You are a grumpy pirate. Always answer in pirate speak."},
        {"role": "user", "content": "How do I write a Python list?"}
    ],
    temperature=0.7
)
print(f"AI Pirate: {role_response.choices[0].message.content}\n")


print("--- 🔢 FEW-SHOT PROMPT ---")
# Few-shot is done by feeding the model fake "history" to establish a strict pattern
few_shot_response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[
        {"role": "system", "content": "You are a helpful assistant that finds exact opposites. Only output the opposite word."},
        # Example 1
        {"role": "user", "content": "Big"},
        {"role": "assistant", "content": "Small"},
        # Example 2
        {"role": "user", "content": "Fast"},
        {"role": "assistant", "content": "Slow"},
        # Example 3
        {"role": "user", "content": "Hot"},
        {"role": "assistant", "content": "Cold"},
        # The Actual Target Prompt
        {"role": "user", "content": "Up"}
    ],
    temperature=0.5
)
print(f"Opposite of Up -> {few_shot_response.choices[0].message.content}")
````
