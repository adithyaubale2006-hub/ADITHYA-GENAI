


### **Prompt engineering**:

> is the process of writing clear, specific instructions to get the best possible output from an AI model.

* **Zero-shot:** Asking the AI to complete a task directly, without giving any examples.
* **Few-shot:** Providing a few examples of the desired input and output to show the AI the exact pattern you want.
* **Role Prompting:** Telling the AI to act as a specific persona (e.g., a senior developer) to guide its tone and expertise level.
```



#### **Linguistics:-**
Linguistics is the study of language

* **PHONETICS:** The study of how speech sounds are produced and perceived
* **PHONOLOGY:** The study of sound patterns and changes
* **MORPHOLOGY:** The study of word structure
* **SYNTAX:** The study of sentence structure
* **SEMANTICS:** The study of linguistic meaning
* **PRAGMATICS:** The study of how language is used in context
* **HISTORICAL:** The study of language change
* **SOCIOLINGUISTICS:** The study of the relation between language and society
* **COMPUTATIONAL:** The study of how computers can process human language
* **PSYCHOLINGUISTICS:** The study of how humans acquire and use language
```
