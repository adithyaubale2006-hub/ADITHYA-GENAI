### ⚙️ **LLM Inference & Decoding**

**LLM Inference** is the operational phase where a trained model is actually put to work generating text, as opposed to the training phase where it learns.

**Key Concepts:**

*   **The Generation Pipeline:** The step-by-step process of inference:
    1.  *Tokenize* the user's text prompt into numbers.
    2.  *Forward Pass:* The model calculates the mathematical probabilities (logits) for what the very next token should be.
    3.  *Decoding:* A specific rule is applied to select the final token from those probabilities.
    4.  *Autoregression:* The chosen token is appended to the input, and the cycle repeats.
*   **Greedy Decoding:** The absolute simplest decoding strategy. The model just picks the #1 most probable next word every single time. It is fast, deterministic, and good for strict factual answers, but it often gets stuck in repetitive loops and sounds robotic.
*   **Sampling:** Instead of strictly picking the top choice, the model rolls a weighted die based on the probabilities. If "blue" has a 70% chance and "clear" has a 20% chance, it will usually pick "blue," but occasionally pick "clear." This is how LLMs achieve natural, creative, and diverse language generation.

```
### 🧠 The Generation Pipeline

Input Prompt: "The cat sat on the"
        │
        ▼
   [ Tokenizer ]
        │
        ▼
[ Large Language Model ]
        │
        ▼
Next Token Probabilities (Logits):
[ mat: 80%  |  rug: 15%  |  dog: 4%  |  moon: 1% ]
        │
        ├───────────────────────────────────────┐
        ▼                                       ▼
 **Greedy Search**                        **Sampling**
(Always takes max)                  (Rolls a weighted die)
        │                                       │
 Picks "mat" (80%)                  Might pick "rug" (15% chance)
```



````python
from transformers import pipeline, set_seed, logging

# Hide warnings
logging.set_verbosity_error()

# Initialize standard GPT-2
generator = pipeline('text-generation', model='gpt2')
prompt = "In a shocking turn of events, the detective discovered that"

print("--- 🤖 GREEDY DECODING ---")
# do_sample=False forces Greedy Search.
# It will always output the exact same text, choosing the #1 probability every time.
greedy_output = generator(prompt, max_new_tokens=40, do_sample=False, pad_token_id=50256)
print(greedy_output[0]['generated_text'] + "\n")


print("--- 🎲 SAMPLING DECODING ---")
# do_sample=True enables probabilistic sampling.
# Combined with temperature and top_p, this stops the model from getting stuck in repetitive loops.
set_seed(42)
sample_output = generator(prompt, max_new_tokens=40, do_sample=True, top_p=0.9, temperature=0.8, pad_token_id=50256)
print(sample_output[0]['generated_text'])
````
