##### **TEXT-EMBEDDINGS**

**Text Embeddings** are a way to translate human language into a format that computers can understand: numbers.

**Key Concepts:**
*   **What is an embedding?** It is a representation of text (a word, a sentence, or a whole document) as a dense vector (a long list of numbers) in a high-dimensional space. The magic of embeddings is that they capture the *semantic meaning* and context of the text, not just the raw characters.
*   **Semantic Similarity:** Because text is mapped into a mathematical space based on its meaning, sentences that mean similar things will have embedding vectors that are mathematically "close" to each other. We can calculate this closeness using metrics like *Cosine Similarity*.
*   **Sentence Transformers:** A powerful and widely used Python library (`sentence-transformers`) built on top of Hugging Face. It provides pre-trained models specifically optimized to generate high-quality embeddings for entire sentences, making it incredibly easy to compare the meanings of different texts.

<br>

##### **SEMANTIC SIMILARITY**

````python
# 1. Install the library (Uncomment the line below and run it once if needed)
# !pip install sentence-transformers

from sentence_transformers import SentenceTransformer, util

# 2. Load a pre-trained sentence transformer model
# 'all-MiniLM-L6-v2' is small, incredibly fast, and excellent for general semantic similarity
model = SentenceTransformer('all-MiniLM-L6-v2')

# 3. Define the sentences to compare
sentence1 = "I am very happy today."
sentence2 = "I feel incredibly joyful right now."
sentence3 = "The weather is terribly gloomy and sad."

# 4. Generate embeddings (convert the text into lists of numbers)
embedding1 = model.encode(sentence1)
embedding2 = model.encode(sentence2)
embedding3 = model.encode(sentence3)

# 5. Calculate and print Cosine Similarity
print("--- Semantic Similarity Scores ---\n")

# Compare Sentence 1 and Sentence 2 (Similar meaning, different words)
sim1_2 = util.cos_sim(embedding1, embedding2)
print(f"Sentence 1: '{sentence1}'")
print(f"Sentence 2: '{sentence2}'")
print(f"Similarity Score: {sim1_2.item():.4f} (Closer to 1 = Highly Similar)\n")

# Compare Sentence 1 and Sentence 3 (Different meaning)
sim1_3 = util.cos_sim(embedding1, embedding3)
print(f"Sentence 1: '{sentence1}'")
print(f"Sentence 3: '{sentence3}'")
print(f"Similarity Score: {sim1_3.item():.4f} (Lower score = Less Similar)")
````
