#### **HUGGING FACE TOKENIZER**

````python
from transformers import AutoTokenizer

#TOKENIZER
tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")

#LOAD THE DATA
input = "Hi iam Adithya Ubale"

#TOKENIZE
tokens = tokenizer(input, return_tensors="pt")

#PRINT
print("TOKENS")
print(tokens)
print()

#PRINT IDS
print("IDS")
print(tokens["input_ids"])
print()

#PRINT ATTENTION MASK
print("ATTENTION MASK")
print(tokens["attention_mask"])
````

<br>

#### **PADDING + TRUNCATION**

````python
# DEFINE A BATCH OF SENTENCES WITH VARYING LENGTHS
batch_sentences = [
    "Hi iam Adithya Ubale",
    "This is short",
    "This is a significantly longer sentence meant to demonstrate exactly how truncation works when we set a maximum token limit."
]

# TOKENIZE WITH PADDING AND TRUNCATION
# padding=True: Adds 0s to shorter sentences so they match the longest one
# truncation=True: Cuts off sentences that exceed the max_length
# max_length=12: Hard limit on the number of tokens
pt_tokens = tokenizer(
    batch_sentences,
    padding=True,
    truncation=True,
    max_length=12,
    return_tensors="pt"
)

# PRINT PADDED AND TRUNCATED IDS
print("PADDED & TRUNCATED IDS:")
print(pt_tokens["input_ids"])
print()

# PRINT ATTENTION MASK
# (Notice how padded tokens get a '0' in the attention mask so the model ignores them)
print("ATTENTION MASK:")
print(pt_tokens["attention_mask"])
````
