### 🗣️ Causal Language Models (Text Generation)

**Causal Language Modeling (CLM)** is the core task behind modern AI text generators. It involves predicting the next word (or token) in a sequence based strictly on the words that came before it.

**Key Concepts:**
*   **Left-to-Right Attention:** The defining feature of a causal model is that it only looks at the *past* (the preceding text) to predict the *future* (the next word) [Causal language modeling - Hugging Face](https://huggingface.co/docs/transformers/v4.35.0/tasks/language_modeling). It cannot look ahead at words that haven't been generated yet. This is often called autoregressive generation.
*   **GPT-Style Models:** Most famous Large Language Models (LLMs) today—like OpenAI's GPT series, Meta's Llama, and the `gpt2` model you used in your code—are causal language models. They are built using "decoder-only" Transformer architectures [Causal language modeling - Hugging Face](https://huggingface.co/docs/transformers/v4.35.0/tasks/language_modeling).
*   **Hugging Face Pipeline:** When you use the `'text-generation'` pipeline in Hugging Face, you are automatically employing a causal language model. The pipeline handles the repetitive loop of predicting the next word, appending it to your original prompt, and predicting the next word again until it reaches your `max_length`.

##### **GPT STYLE HF TEXT-GEN MODEL**

````python
from transformers import pipeline, logging

# Hide extra warnings
logging.set_verbosity_error()

# 1. Initialize the pipeline with GPT-2 (a classic GPT-style model)
gpt_generator = pipeline('text-generation', model='gpt2')

# 2. Experiment with your prompt!
# Causal models are designed to continue whatever text you start.
my_prompt = "In the future, Generative AI will change the way we work by"

# 3. Generate text using sampling parameters
print("--- GPT Text Generation ---\n")
result = gpt_generator(
    my_prompt,
    max_new_tokens=45,   # Limit the length of the generated output
    do_sample=True,      # Tells the model to pick next words probabilistically
    temperature=0.7,     # Controls creativity (Lower = predictable, Higher = random)
    pad_token_id=50256   # Hides the standard GPT-2 padding warning
)

# 4. Print the final combined text
print(result[0]['generated_text'])
````
