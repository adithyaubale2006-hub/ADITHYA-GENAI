#### **TEXT_CLASSIFICATION**

**Text Classification** is a fundamental Natural Language Processing (NLP) task where a model assigns a predefined category or label to a given piece of text.



````python
from transformers import pipeline

text_classifier = pipeline('text-classification')

text = "Hey Hi Iam Adithya I am Learning GEN AI"
print(text_classifier(text))
````

### **Sentiment Analysis**

**Sentiment Analysis** is a specific type of text classification used to determine the emotional tone or attitude expressed in a piece of text.

**Key Concepts:**
*   **Purpose:** To understand if the underlying sentiment of the text is positive, negative, or neutral.
*   **Applications:** Analyzing customer reviews, monitoring social media brand reputation, and understanding user feedback at scale.
*   **How it works in Hugging Face:** By using the `'sentiment-analysis'` pipeline, the library automatically downloads a pre-trained model (usually DistilBERT) fine-tuned specifically to recognize positive and negative emotions.

````python
# 1. Import the pipeline function from the transformers library
from transformers import pipeline

# 2. Initialize the sentiment analysis pipeline
# (Note: 'sentiment-analysis' does the exact same thing as 'text-classification' here)
sentiment_classifier = pipeline('sentiment-analysis')

# 3. Create a list of your own sentences to test
my_sentences = [
    "I absolutely love learning about Generative AI! It's so fascinating.",
    "I'm really frustrated because my code keeps throwing unexpected errors.",
    "The movie was okay, not the best I've seen but not terrible either."
]

# 4. Run the classifier on each sentence and print the results
print("--- Sentiment Analysis Results ---\n")
for sentence in my_sentences:
    result = sentiment_classifier(sentence)
    print(f"Text: {sentence}")
    print(f"Prediction: {result[0]['label']} (Confidence: {result[0]['score']:.4f})\n")
````

###  **Sequence Classification**

**Sequence Classification** is a core NLP task where a model takes an entire sequence of text (a sentence, a paragraph, or a full document) and assigns a single, overall category or label to it.

**Key Concepts:**
*   **Sequence vs. Token:** In sequence classification, the whole input gets *one* label. This is different from *token classification* (like Named Entity Recognition), where every single word or token in the sentence gets its own label.
*   **Broad Applications:** While Sentiment Analysis is the most famous example, sequence classification is also used for Spam Detection (Spam vs. Not Spam), Intent Recognition in chatbots (e.g., classifying "I need a flight" as a `Booking` intent), and Topic Categorization (News vs. Sports vs. Tech).
*   **Under the Hood:** In the Hugging Face `transformers` library, models built for this specific task usually have names ending in `ForSequenceClassification` (e.g., `DistilBertForSequenceClassification`), meaning they have a classification head attached to the end of the base model to output a single label.


````python
# 1. Import the pipeline function
from transformers import pipeline

# 2. Initialize a zero-shot classification pipeline
# (This is a great way to demonstrate sequence classification across custom categories)
sequence_classifier = pipeline("zero-shot-classification")

# 3. Define your text sequence and the categories (labels) you want to test
text_sequence = "I've been spending the last few hours debugging my Python code and reading API documentation."
candidate_labels = ["sports", "technology", "cooking", "travel"]

# 4. Run the classifier
print("--- Sequence Classification Results ---\n")
result = sequence_classifier(text_sequence, candidate_labels)

# 5. Print the top predicted category
print(f"Sequence: '{text_sequence}'")
print(f"Top Prediction: {result['labels'][0]} (Confidence: {result['scores'][0]:.4f})")

# If you want to see all scores for all labels:
print("\nFull breakdown:")
for label, score in zip(result['labels'], result['scores']):
    print(f"- {label}: {score:.4f}")
````
