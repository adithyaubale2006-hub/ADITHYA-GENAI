```
### 1. Hugging Face Datasets & Dataset Loading

The Hugging Face `datasets` library provides access to thousands of ready-to-use community and benchmark datasets for NLP, Audio, and Computer Vision tasks.

Key Features:
* Memory-efficient: Uses Apache Arrow memory-mapping to stream/process massive datasets without running out of RAM.
* Unified API: The `load_dataset()` function loads datasets directly from the Hugging Face Hub, local files (CSV, JSON, Parquet), or cloud storage.
```



````python
# Install datasets and transformers if not already installed
!pip install -q datasets transformers

from datasets import load_dataset

# Using 'stanfordnlp/imdb' to satisfy the namespace/name requirement while loading the correct dataset
dataset = load_dataset("stanfordnlp/imdb")

# Inspect the dataset structure
print("Dataset Structure:")
print(dataset)

# Access a single example from the train set
print("\nFirst sample from training set:")
print(dataset["train"][0])
````

```
### 2. Dataset Splitting (Train / Test / Validation)

Datasets on Hugging Face often come pre-split into `train`, `test`, or `unsupervised`. If a split doesn't exist or you want to create a custom validation split, you can use `.train_test_split()`.

* `test_size`: Specifies the proportion of the dataset to allocate to the test or validation set (e.g., 0.2 = 20%).
* `seed`: Ensures reproducibility across runs.
```



````python
from datasets import load_dataset

# Using 'stanfordnlp/imdb' to satisfy the namespace/name requirement
dataset = load_dataset("stanfordnlp/imdb")

# Take a small slice of the train set to work with quickly
small_train_data = dataset["train"].shuffle(seed=42).select(range(1000))

# Create an 80/20 Train / Validation split
split_data = small_train_data.train_test_split(test_size=0.2, seed=42)

print("Custom Split Structure:")
print(split_data)
print(f"Training samples: {len(split_data['train'])}")
print(f"Validation samples: {len(split_data['test'])}")
````

<br>



```
### 3. Dataset Preprocessing

Before passing text to a language model, it must be preprocessed and converted into numerical token IDs.

Key Preprocessing Functions:
* `.map()`: Applies a transformation function to every row in the dataset in parallel batches.
* `batched=True`: Speeds up processing by handling multiple examples simultaneously.
* `truncation=True` & `padding="max_length"`: Ensures all sequences are uniform in length for the model.
```



````python
from transformers import AutoTokenizer

# 1. Load the tokenizer for the base model
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# 2. Define the tokenization / preprocessing function
def preprocess_function(examples):
    return tokenizer(
        examples["text"],
        padding="max_length",
        truncation=True,
        max_length=128
    )

# 3. Apply the preprocessing across the entire split dataset
tokenized_data = split_data.map(preprocess_function, batched=True)

# 4. View the new features added (input_ids, attention_mask, token_type_ids)
print("\nProcessed Dataset Features:")
print(tokenized_data["train"].column_names)

# Inspect the first tokenized sequence
print("\nSample input_ids (first 10 tokens):")
print(tokenized_data["train"][0]["input_ids"][:10])
````
