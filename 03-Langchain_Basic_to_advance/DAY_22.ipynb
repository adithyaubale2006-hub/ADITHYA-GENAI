#### **04 : APPROACH 1 : ACCESS MODELS HOSTED ON HUGGING FACE THROUGH API**

###### **TEXT2TEXT MODELS | SEQ2SEQ MODELS | ENCODER DECODER MODELS**

````python
from langchain_community.llms import HuggingFaceEndpoint

#Create a Model
llm = HuggingFaceEndpoint(
    repo_id='google/flan-t5-large',
    model_kwargs={'temperature': 0, 'max_length': 64}
)
````

````python
#Create a prompt_template 1
prompt = PromptTemplate(
    input_variable = ["product"],
    template = "What is a good name for Company that makes {product}"
)
````

````python
#Chain the prompt and llm
chain = prompt | llm

chain.invoke({"product": "name"})
````

````python
#Create another prompt_template 2

prompt = PromptTemplate(
    input_variable = ["name"],
    template = "Famous Footballer {name}"
)
````

#### **05 : APPROACH 02 : Download model locally and run it (Create Pipeline)**

##### **Import all the Libraries**

````python
from langchain_community.llms import HuggingFacePipeline
import torch
from transformers import AutoTokenizer, AutoModelForCasualLM, pipeline, AutoModelForSeq2SeqLM
````

````python
#Take A model form HuggingFace
model_id = 'google/flan-t5-large'
````

````python
#Tokenize it
tokenizer = AutoTokenizer.from_pretrained(model_id)

#Model
model = AutoModelForSeq2SeqLM.from_pretrained(model_id)
````

````python
#insert into pipeline
pipeline = pipeline("text2text-generation", model=model, tokenizer=tokenizer, max_length=128)
````

````python
#local_llm using hfPipeline
local_llm = HuggingFacePipeline(pipeline=pipeline)
````

````python
#prompt_template
prompt_template = PromptTemplate(
    input_variable = ["products"],
    template = "What is a good name for Company that make {products}"
)
````

````python
#Chain them
chain = prompt_template | local_llm
````

````python
#print
chain.invoke({"products" : "Delta"})
````
