````python
import os
from langchain_community.document_loaders import PyPDFLoader

# Path to your uploaded PDF file
pdf_path = '/content/MachineLearning.ipynb - Colab.pdf'

if os.path.exists(pdf_path):
    print(f"✅ Found PDF. Loading...")

    # Load the PDF document
    loader = PyPDFLoader(pdf_path)
    documents = loader.load()

    print(f"✅ Successfully loaded {len(documents)} pages from the document.")
else:
    print(f"❌ Error: The file '{pdf_path}' was not found.")
    print("Please check the filename in the sidebar for typos.")
    documents = []
````

#### **CONVERT TO CHUNKS**

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200,
    length_function=len,
)

docs = text_splitter.split_documents(documents)
````

#### **EMBEDDING CONVERSION**

````python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_weaviate import WeaviateVectorStore
import weaviate
import os

# Initialize embeddings using environment variables
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/embedding-001",
    google_api_key = os.environ.get("GOOGLE_API_KEY"),
    max_output_tokens=1024,
)

print("✅ Embeddings initialized successfully.")
````

#### **VECTOR DATABASE STORAGE**

````python
import os
import weaviate
from langchain_weaviate import WeaviateVectorStore
from langchain_google_genai import GoogleGenerativeAIEmbeddings

# 1. Configuration - Clean credentials from environment
# Stripping quotes from URL if they were pasted accidentally
wv_url = os.environ.get("WEAVIATE_URL", "").strip().replace('"', '').replace("https://", "")
wv_key = os.environ.get("WEAVIATE_API_KEY", "").strip().replace('"', '')
gm_key = os.environ.get("GEMINI_API_KEY", "").strip().replace('"', '')

try:
    # 2. Initialize Weaviate client using modern v4 syntax
    client = weaviate.connect_to_weaviate_cloud(
        cluster_url=f"https://{wv_url}",
        auth_credentials=weaviate.auth.AuthApiKey(api_key=wv_key),
        headers={"X-Google-Generative-AI-API-Key": gm_key}
    )

    if client.is_ready():
        print("🚀 Successfully connected to Weaviate Cloud (v4)!")

        # 3. Setup Embeddings (Gemini)
        embeddings = GoogleGenerativeAIEmbeddings(
            model="models/text-embedding-004",
            google_api_key=gm_key
        )

        # 4. Index the 85 chunks
        print(f"Indexing {len(docs)} document chunks...")
        vectorstore = WeaviateVectorStore.from_documents(
            documents=docs,
            embedding=embeddings,
            client=client,
            index_name="MachineLearningDoc"
        )
        print("✅ Vector store created and documents indexed successfully!")
    else:
        print("❌ Connection failed: Cluster not ready.")

except Exception as e:
    print(f"❌ Error: {e}")
finally:
    if 'client' in locals():
        client.close()
        print("Connection closed.")
````

#### **SCHEMA CREATION**

````python
# define input structure
client.schema.delete_all()
client.schema.get()

schema = {
    "classes": [
        {
            "class": "Chatbot",
            "description": "Documents for chatbot",
            "vectorizer": "text2vec-openai",
            "moduleConfig": {"text2vec-openai": {"model": "ada", "type": "text"}},
            "properties": [
                {
                    "dataType": ["text"],
                    "description": "The content of the paragraph",
                    "moduleConfig": {
                        "text2vec-openai": {
                            "skip": False,
                            "vectorizePropertyName": False,
                        }
                    },
                    "name": "content",
                },
            ],
        },
    ]
}

client.schema.create(schema)
vectorstore = WeaviateVectorStore(
    client=client,
    index_name="Chatbot",
    text_key="content",
    embedding=embeddings,
)
````

````python
#Load the text into vectorstore
text_meta_pair = [(doc.page_content, doc.metadata) for doc in docs]
texts, meta = list(zip(*text_meta_pair))
vectorstore.add_texts(text_meta_pair)

"""
Output you might get

['9c496404-7515-4b3c-8b8e-88e2dc09dfc8',
 '0c6883df-8242-4011-a51d-13efd4d6dd39',
 '3ad7f0e4-d5b9-4ffc-9fb0-1d4a4b31f610',
 '4845d89b-4e65-4ec3-ac46-e182793e9b68',
 '32df9a02-fdf1-4e71-81b5-1103264f7499',
 '6e2be198-6bb7-4374-8bf8-bc3f42aefd17',
 '5074cbc0-870c-416e-a37c-63a8f28a06bd',
 '59fd3d13-29f3-4001-af59-7fc48a4e1bec',
 'cb52baa4-bd78-4491-801e-21b074f6aa8d',
 '631ccd1a-e35d-4eb4-94ae-8beb3c1f99e4']

"""
````

### **5. SIMILARITY SEARCH**
Now that the data is indexed, we can perform a similarity search to find the most relevant document chunks based on a query.

````python
query = "What is train test split and why is it important?"

# Perform a basic similarity search
results = vectorstore.similarity_search(query, k=3)

print(f"Top 3 results for query: '{query}'\n")
for i, res in enumerate(results):
    print(f"Result {i+1}:")
    print(f"Content Preview: {res.page_content[:200]}...")
    print(f"Metadata: {res.metadata}\n")
````

### **6. RETRIEVER COMPARISON**
Compare standard retrieval vs. metadata filtering (e.g., searching only within a specific page).

````python
# 1. Search without filters
print("--- Search without filters ---")
search_results = vectorstore.similarity_search("random_state explanation", k=2)
display(search_results[0].page_content[:300])

# 2. Search with metadata filter (example: limit to page 0)
print("\n--- Search with filter (Page 0 only) ---")
filtered_results = vectorstore.similarity_search(
    "random_state explanation",
    k=2,
    filters={"path": ["metadata", "page"], "operator": "Equal", "valueInt": 0}
)

if filtered_results:
    display(filtered_results[0].page_content[:300])
else:
    print("No results found with that specific filter.")
````
