```python
from langchain_community.document_loaders import PyPDFLoader

#Load The File
file_path = "data/TGR.pdf"
pdf = PyPDFLoader(file_path)

data = pdf.load()

data
```

```python
len(data)
```

```python
#Store all the data in a single text file
question_gen = ""
for page in data:
    question_gen += page.page_content + "\n"
```

```python
question_gen
```

```python
from langchain_text_splitters import TokenTextSplitter

#CREATE A SPLITTER TO SPLIT THE DATA INTO CHUNKS OF 10000 TOKENS WITH AN OVERLAP OF 200 TOKENS
splitter_ques_gen = TokenTextSplitter(
    chunk_size = 30000,
    chunk_overlap = 200
)

#CHUNKS use split_text
chunks_ques_gen = splitter_ques_gen.split_text(question_gen)
```

```python
chunks_ques_gen
```

```python
len(chunks_ques_gen)
```

```python
#Now we convert this into a document format so that we can use it in the vector store
from langchain_community.docstore.document import Document

document_ques_gen = [Document(page_content=chunk) for chunk in chunks_ques_gen]
```

```python
document_ques_gen #Converted into document format
```

```python
splitter_ans_gen = TokenTextSplitter(
    chunk_size = 30000, 
    chunk_overlap = 200
)
```

```python
#Use split_documents to split the document into chunks of 10000 tokens with an overlap of 200 tokens
document_answer_gen = splitter_ans_gen.split_documents(document_ques_gen) 
```

```python
document_answer_gen
```

```python
len(document_answer_gen)
```

```python
from langchain_nvidia import ChatNVIDIA

#Create a LLM model using NVIDIA's API
llm = ChatNVIDIA(
    model = "openai/gpt-oss-20b",
    api_key = os.environ["NVIDIA_API_KEY"],
    temperature = 0.5,
    max_tokens = 1024,
    top_p = 1,
    timeout=200
)
```

```python
response = llm.invoke("Hello, are you working?")
print(response.content)
```

```python
prompt_template = """
You are an expert personal growth coach, mindset strategist, and master examiner.
Your goal is to deeply analyze the principles, laws of success, and mental frameworks from the text below and test a reader's absolute comprehension of them:

-----------
{text}
-----------

Create challenging, thought-provoking exam-style questions based strictly on the text provided. 
Focus on key concepts like the subconscious mind, definite major purpose, autosuggestion, persistence, transmutation, and psychological laws. 
Make sure not to lose any core philosophical or practical success principles.

QUESTIONS:
"""
```

```python
from langchain_core.prompts import PromptTemplate

PROMPT_QUESTIONS = PromptTemplate(
    input_variables=["text"],
    template=prompt_template
)
```

```python
refine_template = ("""
You are an expert personal growth coach, mindset strategist, and master examiner.
Your goal is to help a reader test their mastery of the principles of success, wealth-building laws, and mental frameworks.
We have received some practice questions to a certain extent: {existing_answer}.
We have the option to refine the existing questions or add new ones 
(only if necessary) with some more context below.

{text}
------------
Given the new context, refine the original questions to better target core philosophical concepts like definite purpose, autosuggestion, and the subconscious mind.
If the context is not helpful, please provide the original questions.
QUESTIONS:
""")
```

```python
REFINE_PROMPT_QUESTIONS = PromptTemplate(
    input_variables=["existing_answer", "text"],
    template=refine_template
)
```

```python
from langchain_classic.chains.summarize import load_summarize_chain

ques_chain_gen = load_summarize_chain(
    llm, 
    chain_type="refine", 
    question_prompt=PROMPT_QUESTIONS, 
    refine_prompt=REFINE_PROMPT_QUESTIONS, 
    verbose=True
)

ques = ques_chain_gen.invoke(document_answer_gen)

print(ques["output_text"])
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

# Initialize a free, fast, locally-run embedding model
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# Build your FAISS vector store
vector_stores = FAISS.from_documents(document_answer_gen, embeddings)
```

```python
llm = ChatNVIDIA(
    model = "openai/gpt-oss-20b",
    api_key = os.environ["NVIDIA_API_KEY"],
    temperature = 0.1,
    max_tokens = 1024,
    top_p = 1,
    timeout = 200
)
```

```python
# Split the text into a list of lines
raw_questions = ques["output_text"].split("\n")

# Keep only the lines that actually contain text (removes empty lines)
clean_questions = [q.strip() for q in raw_questions if q.strip()]

# Print exactly the first 10 clean questions
for i, question in enumerate(clean_questions[:10]):
    print(f"Question {i+1}: {question}")

    
```

```python
from langchain_classic.chains import RetrievalQA

generation_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=vector_stores.as_retriever()
)
```

```python
import time

# 1. Split and clean the questions
raw_questions = ques["output_text"].split("\n")
clean_questions = [q.strip() for q in raw_questions if q.strip()]

# 2. Loop through the clean questions with rate-limiting and error handling
for question in clean_questions[:4]:  # Limits to the first 10 clean questions
    if not question.strip():
        continue
        
    print(f"Question: {question}")
    
    try:
        # Get the response using your RetrievalQA chain
        response = generation_chain.invoke(question) 
        answer = response["result"] 
        
        print(f"Answer: {answer}")
        print("-" * 50)
        
        # Save to file
        with open("answers.txt", "a", encoding="utf-8") as f:
            f.write(f"Question: {question}\n")
            f.write(f"Answer: {answer}\n")
            f.write("-" * 50 + "\n")
            
        # Pause for 3 seconds to avoid triggering API timeouts/rate limits
        time.sleep(3) 
        
    except Exception as e:
        print(f"Failed to answer this question due to error: {e}")
        print("-" * 50)
        
        # Pause a bit longer if an error occurs before continuing
        time.sleep(5)
```
